```python
import matplotlib.pyplot as plt
import geopandas as gpd
from shapely.geometry import box
import contextily as cx

# Coordinate centrali
center_lon, center_lat = 11.1862, 44.0853

# Crea bounding box: +/- 0.01 gradi (circa 1 km)
minx, miny = 10.734269464357556, 43.96629819030445
maxx, maxy = 12.139133497965453, 44.91004596649926

# Crea poligono bbox
bbox = box(minx, miny, maxx, maxy)
gdf = gpd.GeoDataFrame(geometry=[bbox], crs="EPSG:4326")

# Converti in Web Mercator per contextily
gdf_web = gdf.to_crs(epsg=3857)

# Plot
fig, ax = plt.subplots(figsize=(8, 8))
gdf_web.boundary.plot(ax=ax, edgecolor='red', linewidth=2)
cx.add_basemap(ax, source=cx.providers.OpenStreetMap.Mapnik)
ax.set_title("Bounding Box around (11.1862, 44.0853)")
plt.axis("off")
plt.tight_layout()
plt.show()
```
