## Use OTP

```python
import digitalhub as dh

from otp_mobility.utils.constants import PROJECT
from config import params_list
```

```python
# Project
project = dh.get_project(PROJECT)
```

```python
# Define the main function
otp_func = project.new_function("otp-processor",
                                    kind="python",
                                    python_version="PYTHON3_10",
                                    code_src="processor.py",
                                    handler="main_platform")
```

```python
# Define helper functions to manage parameters
# !!!Go to the DHCore platform, get the key of input data and service url of running containers!!!
def get_additional_params(method, zoi, modes):
    if method == "simplified":
        od_input_key = "store://test-otp-v2/dataitem/table/od-coords-simplified:7458a32abd9d4a71af92653add14a2fe"
    if method == "extended":
        od_input_key = "store://test-otp-v2/dataitem/table/od-coords-extended:5debda04f67a4741a0cb8f9ad36f934c"
    if method == "av":
        od_input_key = "store://test-otp-v2/dataitem/table/od-coords-av:8d3edc29f83347e48a5669870be67625"

    modes_str =  "_".join(modes)
    output_times_file = f"otp_results_{method}_{zoi}_{modes_str}"
    
    if zoi == "allBologna":
        url_port = "s-containerserve-f3023cd94f6d4b3b8542b89f6a981735.digitalhub-bologna:8080"
    elif zoi == "restrictedAv":
        url_port = "s-containerserve-76479900e00f47ee9b377e8f6a62443d.digitalhub-bologna:8080"
    otp_endpoint = f"http://{url_port}/otp/gtfs/v1"

    return od_input_key, output_times_file, otp_endpoint

def check_params(params):
    if (params["method"] not in ["extended", "simplified", "av"]) or (params["zoi"] not in ["allBologna", "restrictedAv"]):
        return False
    for mode in params["modes"]:
        if mode not in ["CAR", "WALK", "CAR_PARK", "TRANSIT"]:
            return False
    return True
                
```

```python
# Run the simulations
# for all combinations of parameters
for params in params_list:
    if not check_params(params):
        print(f"The parameters provided are: {params}. They are not acceptable")
        continue

    method = params["method"]
    zoi = params["zoi"]
    modes = params["modes"]

    od_input_key, output_times_file, otp_endpoint = get_additional_params(method, zoi, modes)
    
    otp_run = otp_func.run("job", 
                       inputs={'df': od_input_key},
                       parameters={'main_modes': modes,
                                   'output_name': output_times_file,
                                   'otp_endpoint': otp_endpoint})
```
